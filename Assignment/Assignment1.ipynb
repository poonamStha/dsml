```python
# 1st
txt = "Hello! Python World"
print(txt.upper())
print(txt.split())
print(txt.lower())
print(txt.capitalize())
print(txt.isalpha())
print(txt.isalnum())
print(txt.index('Python'))
print(txt.find('World'))
print(txt.isdigit())
print(txt.replace('Python',"DataScience"))
```

```python
# 2nd
# taking sentence from user
txt = input("Enter a sentence:")
# converting case
print("Uppercase:",txt.upper())
print("Lowercase:",txt.lower())
# counting words
print("length:",len(txt))
# checking digits/alphabet
print("indexing:", txt.find('a'))
print(txt.isalpha())
print(txt.isalnum())

# replacing words
old = input("Enter a word:")
new = input("Enter a replacing word:")
print("replaced:",txt.replace(old,new))
```
