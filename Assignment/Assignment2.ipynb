```python
def student_result():
  students = {}
  marks_list = []

  for i in range(5):
    name = input("Enter name:")
    marks = float(input("Enter obtain marks:"))
    students[name]=marks
    marks_list.append(marks)

  for name in students:
    marks = students[name]

    if marks >= 80:
      grade = 'A'
    elif marks >= 60:
      grade = 'B'
    elif marks >= 40:
      grade = 'C'
    else:
      grade = 'F'
    print(name,":",marks,"-",grade)
  print("Highest:",max(marks_list))
  print("Lowest:",min(marks_list))
  print("Average:",sum(marks_list)/5)
student_result()
```

```python
def shopping_cart():
  product_price = []

  for i in range(5):
    price = float(input("Enter a price:"))
    product_price.append(price)

  total = 0
  for p in product_price:
    total += p

  if total >= 5000:
    discount = total *0.15
  elif total >= 3000:
    discount = total * 0.10
  elif total >= 1000:
    discount = total *0.05
  else:
    discount = 0

  final_amount = total - discount

  print("Products:",product_price)
  print("Total_amount:", total)
  print("Discount_amount:",discount)
  print("Final_amount:",final_amount)
shopping_cart()
```
