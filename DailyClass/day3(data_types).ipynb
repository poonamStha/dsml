**Python Numerical Types**


*   **Integers**: Integers are whole numbers,either positive or negative. In python,integers have no size limit. It represent using int type
*   **Floating_point Numbers**: Floating-point numbers are decimal numbers,either positive or negative. In python, Floting- point numbers represent using float type
* **complex numbers**: Complex numbers are numbers with a real part and imaginary part. In python, complex numbers represent using complex type.
