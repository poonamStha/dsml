**Deep copy and Shallow copy**
images.png

```python
lst = [1,2,3,4]
lst1 = lst
print(lst)
print(lst1)
```

```python
lst2 = lst1.copy()# shallow copy
lst2
```

```python
lst = [[1,2,3,4],[5,6,7]]
lst1 = lst.copy()
lst[1][1]=100
print(lst)
print(lst1)
```

```python
# eg of deep copy
import copy
lst = [[1,2,3,4],[5,6,7]]
lst1 = copy.deepcopy(lst)
lst[1][2]=100
print(lst)
print(lst1)
```
