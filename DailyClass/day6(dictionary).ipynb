**join two dictionary**


*   Using update()
*   using **kwargs to unpack



```python
# eg of update()
dict1 = {'name':"poona"}
dict2 = {'roll no':1}
dict1.update(dict2)
dict1
```

```python
# eg of using **kwargs
dict1 = {'name':"poona"}
dict2 = {'roll no':1}
dict3 = {'age':25}
dict4 = {**dict1, **dict2, **dict3}
dict4
```
