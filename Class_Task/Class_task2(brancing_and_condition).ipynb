```python
grade = float(input("enter your grade:"))
if grade < 1.6:
  print("NG")
elif (grade >= 1.6 and grade < 2.0):
  print("basic")
elif (grade >= 2.0 and grade < 2.4):
  print("acceptable")
elif (grade >= 2.4 and grade < 2.8):
  print("satisfactory")
elif (grade >= 2.8 and grade < 3.2):
  print("good")
elif (grade >= 3.2 and grade < 3.6):
  print("excellent")
else:
  print("outstanding")
```
